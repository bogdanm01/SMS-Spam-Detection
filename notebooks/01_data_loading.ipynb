# Prvo učitavanje SMS podataka

Ova sveska učitava originalni Kaggle fajl bez čišćenja. Cilj je da vidimo kolone, broj redova i oznake. Tekst poruka se u ovoj celini ne menja.

```python
from pathlib import Path
import sys

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.data_loading import load_raw_sms

sms = load_raw_sms()
print("Redovi i kolone:", sms.shape)
print("Kolone:", sms.columns.tolist())
```

`shape` vraća `(broj_redova, broj_kolona)`. `v1` je oznaka klase, a `v2` prvi deo teksta. Dodatne kolone zadržavamo jer u nekim redovima sadrže nastavak poruke.

```python
sms[["v1", "v2"]].head(3)
```

```python
print("Broj poruka po klasi:")
print(sms["v1"].value_counts())
print("Redovi sa sadržajem u dodatnim kolonama:", sms.iloc[:, 2:].notna().any(axis=1).sum())
```

Ovo su podaci pre čišćenja. Naredna celina ispitaće strukturu problematičnih redova, nedostajuće vrednosti i duplikate, pa ćemo tek onda odrediti pravila obrade.
